### Deployment Description

The model will help doctors identify whether a breast tumour is benign or malignant using diagnostic measurements. It will make predictions whenever new test results are available and return them within a few seconds. Each prediction will include an explanation of the important features, helping doctors understand the result. The model will support, not replace, doctors' decisions.

**Primary Metric: Recall for the Malignant Class**

Recall for the malignant class is selected as the primary evaluation metric because missing a malignant case can have serious consequences in a medical decision-support setting. Recall measures the proportion of actual malignant cases correctly identified by the model. A high recall reduces the number of false negatives. Accuracy will also be reported, but recall is more important for the primary objective of identifying malignant cases. The model's predictions would still require professional medical review.

## R2

```python
from sklearn.preprocessing import StandardScaler
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=42,test_size=0.20,stratify=y)
```

```python
from sklearn.model_selection import StratifiedKFold

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC

knn_pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("model", KNeighborsClassifier())
])

nb_pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("model", GaussianNB())
])

svm_pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("model", SVC())
])
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import make_scorer, recall_score

malignant_recall = make_scorer(
    recall_score,
    pos_label="M"
)

param_grids = {
    "KNN": {
        "model__n_neighbors": [3, 5, 7, 9, 11]
    },
    "GaussianNB": {
        "model__var_smoothing": [
            1e-11, 1e-10, 1e-9, 1e-8, 1e-7
        ]
    },
    "SVM": {
        "model__C": [0.1, 1, 10, 100, 1000]
    }
}

pipelines = {
    "KNN": knn_pipe,
    "GaussianNB": nb_pipe,
    "SVM": svm_pipe
}

searches = {}

for name, pipe in pipelines.items():
    search = GridSearchCV(
        estimator=pipe,
        param_grid=param_grids[name],
        cv=cv,
        scoring=malignant_recall,
        n_jobs=-1
    )

    search.fit(X_train, y_train)
    searches[name] = search

    print(name)
    print("Best parameters:", search.best_params_)
    print("Best CV malignant recall:", round(search.best_score_, 4))
    print()
```

```python
from sklearn.metrics import accuracy_score, recall_score, f1_score


results = []

for name, search in searches.items():
    y_pred = search.predict(X_test)

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Malignant Recall": recall_score(
            y_test, y_pred, pos_label="M"
        ),
        "Malignant F1": f1_score(
            y_test, y_pred, pos_label="M"
        )
    })

results_df = pd.DataFrame(results)
print(results_df.round(4))
```

## R4

```python
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import loguniform
```

```python
from sklearn.metrics import make_scorer, recall_score

malignant_recall = make_scorer(
    recall_score,
    pos_label="M"
)
```

```python
from sklearn.model_selection import StratifiedKFold, RandomizedSearchCV
from scipy.stats import randint, loguniform


cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


knn_pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("model", KNeighborsClassifier())
])

knn_params = {
    "model__n_neighbors": randint(3, 31),
    "model__weights": ["uniform", "distance"],
    "model__metric": ["euclidean", "manhattan"]
}

knn_search = RandomizedSearchCV(
    estimator=knn_pipe,
    param_distributions=knn_params,
    n_iter=20,
    scoring=malignant_recall,
    cv=cv,
    random_state=42,
    n_jobs=-1,
    return_train_score=False
)

knn_search.fit(X_train, y_train)

print("Best KNN parameters:", knn_search.best_params_)
print("Best CV recall:", knn_search.best_score_)
```

```python
from scipy.stats import loguniform

nb_pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("model", GaussianNB())
])

nb_params = {
    "model__var_smoothing": loguniform(1e-12, 1e-6)
}

nb_search = RandomizedSearchCV(
    estimator=nb_pipe,
    param_distributions=nb_params,
    n_iter=20,
    scoring=malignant_recall,
    cv=cv,
    random_state=42,
    n_jobs=-1,
    return_train_score=False
)

nb_search.fit(X_train, y_train)

print("Best Naive Bayes parameters:", nb_search.best_params_)
print("Best CV malignant recall:", nb_search.best_score_)
```

```python
from scipy.stats import loguniform, randint

svm_pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("model", SVC())
])

svm_params = [
    {
        "model__kernel": ["rbf"],
        "model__C": loguniform(1e-3, 1e3),
        "model__gamma": loguniform(1e-5, 1e1)
    },
    {
        "model__kernel": ["linear"],
        "model__C": loguniform(1e-3, 1e3),
        "model__gamma": ["scale"]
    }
]

svm_search = RandomizedSearchCV(
    estimator=svm_pipe,
    param_distributions=svm_params,
    n_iter=20,
    scoring=malignant_recall,
    cv=cv,
    random_state=42,
    n_jobs=-1,
    return_train_score=False
)

svm_search.fit(X_train, y_train)

print("Best SVM parameters:", svm_search.best_params_)
print("Best CV malignant recall:", svm_search.best_score_)
```

```python

searches = {
    "KNN": knn_search,
    "GaussianNB": nb_search,
    "SVM": svm_search
}

summary = []

for name, search in searches.items():
    scores = np.asarray(search.cv_results_["mean_test_score"])
    best_score = search.best_score_
    score_std = np.std(scores, ddof=1)

    within_one_std = np.sum(
        scores >= best_score - score_std
    )

    summary.append({
        "Model": name,
        "Best CV Recall": best_score,
        "CV Score Std": score_std,
        "Candidates Tested": len(scores),
        "Within 1 Std of Best": within_one_std
    })

r4_summary = pd.DataFrame(summary)
print(r4_summary.round(4))
```

## R5

```python
from sklearn.model_selection import RepeatedStratifiedKFold

cv_r5 = RepeatedStratifiedKFold(
    n_splits=5,
    n_repeats=5,
    random_state=42
)
```

```python
from sklearn.base import clone
from sklearn.model_selection import cross_validate
from sklearn.metrics import make_scorer, recall_score

malignant_recall = make_scorer(
    recall_score,
    pos_label="M"
)

best_models = {
    "KNN": knn_search.best_estimator_,
    "GaussianNB": nb_search.best_estimator_,
    "SVM": svm_search.best_estimator_
}

r5_scores = {}

for name, model in best_models.items():
    result = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv_r5,
        scoring=malignant_recall,
        n_jobs=-1
    )

    r5_scores[name] = result["test_score"]

    print(name, "Mean recall:", round(result["test_score"].mean(), 4))
```

```python
from itertools import combinations

paired_results = []

for model_a, model_b in combinations(r5_scores.keys(), 2):
    scores_a = r5_scores[model_a]
    scores_b = r5_scores[model_b]

    differences = scores_a - scores_b

    paired_results.append({
        "Model A": model_a,
        "Model B": model_b,
        "Mean Difference (A-B)": differences.mean(),
        "Difference Std": differences.std(ddof=1),
        "A Wins": np.sum(differences > 0),
        "B Wins": np.sum(differences < 0),
        "Ties": np.sum(differences == 0)
    })

paired_df = pd.DataFrame(paired_results)
print(paired_df.round(4))
```

### R5 — Paired Comparison and Limitations

The three models were compared using the same repeated stratified cross-validation folds and malignant-class recall. The mean paired difference shows which model performed better on average, while the standard deviation shows how much the difference varied across folds. The win counts show how often one model achieved higher recall than another.

These results describe model performance on the selected dataset, but they do not prove that one model is statistically superior. First, the same dataset was used for hyperparameter tuning and model comparison, which may lead to optimistic results. Second, repeated cross-validation folds overlap and are not independent observations. Therefore, the differences should be interpreted with caution.

## R6

```python
from sklearn.metrics import roc_auc_score,confusion_matrix
best_models = {
    "KNN": knn_search.best_estimator_,
    "GaussianNB": nb_search.best_estimator_,
    "SVM": svm_search.best_estimator_
}

r6_results = []

for name, model in best_models.items():
    y_pred = model.predict(X_test)

    recall = recall_score(
        y_test, y_pred, pos_label="M"
    )

    cm = confusion_matrix(
        y_test, y_pred, labels=["B", "M"]
    )
    tn, fp, fn, tp = cm.ravel()

    if hasattr(model, "predict_proba"):
        scores = model.predict_proba(X_test)[:, list(model.classes_).index("M")]
    else:
        scores = model.decision_function(X_test)
        if list(model.classes_).index("M") == 0:
            scores = -scores

    auc = roc_auc_score(
        (y_test == "M").astype(int), scores
    )

    r6_results.append({
        "Model": name,
        "Malignant Recall": recall,
        "ROC-AUC": auc,
        "Accuracy": accuracy_score(y_test, y_pred),
        "False Negatives": fn,
        "Confusion Matrix": cm.tolist()
    })

r6_df = pd.DataFrame(r6_results)
print(r6_df.to_string(index=False))
```

```python
cv_scores = {
    "KNN": knn_search.best_score_,
    "GaussianNB": nb_search.best_score_,
    "SVM": svm_search.best_score_
}

r6_df["CV Malignant Recall"] = r6_df["Model"].map(cv_scores)
r6_df["Test-CV Gap"] = (
    r6_df["Malignant Recall"] - r6_df["CV Malignant Recall"]
)

print(
    r6_df[
        ["Model", "CV Malignant Recall",
         "Malignant Recall", "Test-CV Gap"]
    ].round(4).to_string(index=False)
)
```

## R7

```python

cost_rows = []

for name, model in best_models.items():
    clf = model.named_steps["model"]

    if name == "KNN":
        stored_numbers = (
            clf._fit_X.size + len(clf._y)
        )
        work = len(clf._fit_X) * clf.n_features_in_

    elif name == "GaussianNB":
        stored_numbers = (
            clf.theta_.size + clf.var_.size
            + clf.class_prior_.size
        )
        work = len(clf.theta_) * clf.n_features_in_

    else:
        stored_numbers = (
            clf.support_vectors_.size
            + clf.dual_coef_.size
            + clf.intercept_.size
        )
        work = (
            len(clf.support_vectors_) * clf.n_features_in_
        )

    cost_rows.append({
        "Model": name,
        "Stored Numbers (approx.)": stored_numbers,
        "Work per Prediction (approx.)": work
    })

print(pd.DataFrame(cost_rows).to_string(index=False))
```

```python
import numpy as np
import pandas as pd

# Choose one test case
case_idx = 0
x_case = X_test.iloc[[case_idx]]

knn_model = best_models["KNN"]
knn_clf = knn_model.named_steps["model"]
scaler = knn_model.named_steps["scaler"]

X_train_scaled_knn = scaler.transform(X_train)
x_case_scaled = scaler.transform(x_case)

distances = np.sqrt(
    ((X_train_scaled_knn - x_case_scaled[0]) ** 2).sum(axis=1)
)

k = knn_clf.n_neighbors
nearest_idx = np.argsort(distances)[:k]

nearest_cases = pd.DataFrame({
    "Distance": distances[nearest_idx],
    "Actual Label": y_train.iloc[nearest_idx].to_numpy()
})

print("Selected test case actual label:", y_test.iloc[case_idx])
print("KNN predicted label:", knn_model.predict(x_case)[0])
print(nearest_cases)
```

```python

nb_model = best_models["GaussianNB"]
nb_clf = nb_model.named_steps["model"]
nb_scaler = nb_model.named_steps["scaler"]

x_case_scaled = nb_scaler.transform(X_test.iloc[[case_idx]])[0]


classes = nb_clf.classes_
contributions = []

for i, cls in enumerate(classes):
    log_likelihood = -0.5 * (
        np.log(2 * np.pi * nb_clf.var_[i])
        + ((x_case_scaled - nb_clf.theta_[i]) ** 2) / nb_clf.var_[i]
    )

    contributions.append(log_likelihood)

contrib_df = pd.DataFrame(
    np.array(contributions).T,
    columns=[f"Class_{cls}" for cls in classes],
    index=X.columns
)

print("Actual label:", y_test.iloc[case_idx])
print("NB predicted label:", nb_model.predict(X_test.iloc[[case_idx]])[0])
print(contrib_df.round(3))
```

```python

svm_model = best_models["SVM"]

x_case = X_test.iloc[[case_idx]]
svm_prediction = svm_model.predict(x_case)[0]
svm_score = svm_model.decision_function(x_case)[0]

print("Actual label:", y_test.iloc[case_idx])
print("SVM predicted label:", svm_prediction)
print("SVM decision score:", round(float(svm_score), 4))
print("Kernel:", svm_model.named_steps["model"].kernel)
```

```python
from sklearn.metrics import log_loss
import numpy as np
import pandas as pd

calibration_results = []

for name, model in best_models.items():
    # Get probability estimates
    if hasattr(model, "predict_proba"):
        probabilities = model.predict_proba(X_test)
        classes = list(model.classes_)
        malignant_idx = classes.index("M")
        p_malignant = probabilities[:, malignant_idx]
    else:
        # Convert SVM decision scores to probabilities
        # Requires SVC(probability=True) to be fitted
        print(name, "does not provide predict_proba; refit SVM with probability=True.")
        continue

    y_true = (np.asarray(y_test) == "M").astype(int)
    y_pred = (p_malignant >= 0.5).astype(int)

    extreme = (p_malignant > 0.99) | (p_malignant < 0.01)
    wrong_extreme = np.sum(y_pred[extreme] != y_true[extreme])

    calibration_results.append({
        "Model": name,
        "Log Loss": log_loss(y_true, p_malignant),
        "Extreme Predictions": int(extreme.sum()),
        "Extreme Share": float(extreme.mean()),
        "Wrong Extreme Predictions": int(wrong_extreme)
    })

print(pd.DataFrame(calibration_results).round(4).to_string(index=False))
```
